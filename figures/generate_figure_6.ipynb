# 🖼️ Generating Figure 6

The following code generates Figure 6 of the paper by plotting the performance of the unconditioned model against the conditioned model for SatBird and sPlotOpen when varying the size of the training data.

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
plt.rcParams.update({'font.size': 10.3})

ciso = np.array([1, 5, 10, 20, 50, 100])

non_song_uncond = np.array([48.69, 55.18, 55.98, 56.98, 58.84, 59.41])
non_song_cond = np.array([48.31, 56.60, 58.37, 60.46, 62.61, 63.56])

song_uncond = np.array([57.02, 64.41, 65.79, 67.01, 68.49, 69.30])
song_cond = np.array([57.20, 66.19, 68.30, 70.33, 71.94, 72.86])

non_tree_uncond = np.array([74.10, 92.63, 94.21, 95.28, 96.07, 96.49])
non_tree_cond = np.array([73.46, 92.97, 94.93, 96.26, 96.98, 97.39])

tree_uncond = np.array([82.57, 95.89, 96.75, 97.73, 98.24, 98.49])
tree_cond = np.array([81.64, 96.26, 97.62, 98.46, 98.93, 99.12])

fig, axes = plt.subplots(1, 4, figsize=(12, 3), sharey=False)

# -------------------------
# Plot 1: Songbirds
# -------------------------
axes[0].plot(ciso, song_uncond, marker='s', label="Unconditioned", linestyle='--', color=sns.color_palette("Paired")[9])
axes[0].plot(ciso, song_cond, marker='o', label="Conditioned", color=sns.color_palette("Paired")[9])

axes[0].set_title("Songbirds")
axes[0].set_xlabel("Fraction of training data (%)")

axes[0].set_ylabel("Top-$k$ (%)")
#axes[1].set_xlim(0, 100)

axes[0].legend()
axes[0].grid(True)

# -------------------------
# Plot 2: Non-songbirds
# -------------------------
# Make uncond dashed
axes[1].plot(ciso, non_song_uncond, marker='s', label="Unconditioned", linestyle='--', color=sns.color_palette("Paired")[8])
axes[1].plot(ciso, non_song_cond, marker='o', label="Conditioned", color=sns.color_palette("Paired")[8])

axes[1].set_title("Non-Songbirds")
axes[1].set_xlabel("Fraction of training data (%)")

axes[1].set_ylabel("Top-$k$ (%)")
#axes[1].set_xlim(0, 100)
axes[1].legend()
axes[1].grid(True)

# -------------------------
# Plot 3: Trees
# -------------------------
axes[2].plot(ciso, tree_uncond, marker='s', label="Unconditioned", linestyle='--', color=sns.color_palette("Paired")[2])
axes[2].plot(ciso, tree_cond, marker='o', label="Conditioned", color=sns.color_palette("Paired")[2])

axes[2].set_title("Trees")
axes[2].set_xlabel("Fraction of training data (%)")
axes[2].set_ylabel("AUC (%)")
#axes[1].set_xlim(0, 100)
axes[2].legend()
axes[2].grid(True)

# -------------------------
# Plot 4: Non-trees
# -------------------------
axes[3].plot(ciso, non_tree_uncond, marker='s', label="Unconditioned", linestyle='--', color=sns.color_palette("Paired")[3])
axes[3].plot(ciso, non_tree_cond, marker='o', label="Conditioned", color=sns.color_palette("Paired")[3])

axes[3].set_title("Non-Trees")
axes[3].set_xlabel("Fraction of training data (%)")
#axes[3].set_xticks([1, 5, 10, 20, 50, 100])
#axes[1].set_xlim(0, 100)
axes[3].legend()
axes[3].grid(True)

axes[3].set_ylabel("AUC (%)")

plt.tight_layout()
# reduce space between subplots
plt.subplots_adjust(wspace=0.34)

# reduce space on the left and right
plt.subplots_adjust(left=0.053, right=0.996, top=0.92, bottom=0.145)
plt.savefig("training_data_fraction.pdf", dpi=300)
plt.show()
```
